## Fundamentos teóricos

A radiação térmica corresponde à emissão de energia eletromagnética associada à temperatura absoluta da matéria. Diferentemente da condução e da convecção, não requer meio material para ocorrer.

### Corpo negro

O corpo negro é um emissor ideal que absorve toda a radiação incidente e emite a máxima quantidade possível de energia para uma dada temperatura.

A potência emissiva espectral hemisférica é dada pela Lei de Planck:

$$
E_\lambda^b(\lambda,T)=\frac{2\pi h c^2}{\lambda^5}
\frac{1}{\exp\left(\frac{hc}{\lambda k_B T}\right)-1}.
$$

A emissão total resulta da integração espectral:

$$
E_b = \sigma T^4,
$$

enquanto o comprimento de onda correspondente ao pico de emissão obedece à lei de Wien:

$$
\lambda_{pico} T = b.
$$

### Superfícies reais (corpo cinzento)

Superfícies reais emitem menos energia do que um corpo negro à mesma temperatura. Para uma superfície difusa-cinzenta, define-se a emissividade $\varepsilon$ tal que

$$
E = \varepsilon \sigma T^4.
$$

## Modelo matemático da troca radiativa

### Superfície trocando com envoltória grande

Quando uma superfície troca radiação com uma envoltória grande e isotérmica, o fluxo líquido de calor por radiação é

$$
q_{rad} = \varepsilon \sigma A \left(T_s^4 - T_{sur}^4\right).
$$

Para pequenas diferenças de temperatura, essa expressão pode ser linearizada na forma

$$
q_{rad} \approx h_r A (T_s - T_{sur}), \qquad h_r = 4\varepsilon\sigma T_m^3.
$$

### Duas superfícies difusas-cinzentas com $F_{12}=1$

Em configurações geométricas onde todo o fluxo radiativo emitido por uma superfície atinge a outra, o balanço radiativo pode ser representado por uma rede de resistências.

A taxa líquida de transferência de calor é

$$
q = \frac{\sigma (T_1^4 - T_2^4)}{\frac{1-\varepsilon_1}{A\varepsilon_1}
+ \frac{1}{A} + \frac{1-\varepsilon_2}{A\varepsilon_2}}
= \varepsilon_{eff} \sigma A (T_1^4 - T_2^4),
$$

onde a emissividade efetiva é definida por

$$
\varepsilon_{eff} = \left( \frac{1}{\varepsilon_1} + \frac{1}{\varepsilon_2} - 1 \right)^{-1}.
$$
