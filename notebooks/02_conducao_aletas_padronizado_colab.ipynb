# Notebook 2 — Condução com Superfícies Estendidas (Aletas)

**Objetivo.** Analisar a transferência de calor por condução em superfícies estendidas (aletas), avaliando a influência de parâmetros geométricos, propriedades térmicas e do número de aletas sobre o desempenho térmico do sistema.

## Hipóteses físicas e escopo

- Regime permanente.
- Propriedades térmicas constantes.
- Condução unidimensional ao longo da aleta.
- Coeficiente convectivo uniforme.
- Radiação desprezível.

Essas hipóteses permitem empregar soluções analíticas clássicas para a temperatura na aleta.

## Fundamentos teóricos

Superfícies estendidas são utilizadas para aumentar a taxa de transferência de calor entre um sólido e um fluido, ampliando a área disponível para convecção.

O comportamento térmico da aleta resulta do balanço entre condução ao longo do sólido e convecção na superfície lateral.

## Modelo matemático

Para uma aleta de seção constante em regime permanente, sem geração interna, a equação governante é

$$
\frac{d^2\theta}{dx^2} - m^2\theta = 0,
$$

onde $\theta(x)=T(x)-T_\infty$ e

$$
m^2 = \frac{hP}{kA_c}.
$$

## Discussão física

Os resultados mostram que a eficiência da aleta diminui à medida que seu comprimento aumenta, devido ao aumento da resistência condutiva.

O aumento do número de aletas amplia a taxa total de transferência de calor, porém com ganhos marginais decrescentes, associados à interferência térmica e ao limite imposto pelo coeficiente convectivo.
